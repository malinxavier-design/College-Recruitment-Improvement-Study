## 4. Data Cleaning & Transformation

### 4.1 Renaming Columns
The raw column headers are full survey questions with brackets, special characters, and trailing spaces (e.g., `'Age '`). We rename key columns to concise, lowercase words using underscores for clarity.

```python

df.columns = df.columns.str.strip()


rename_dict = {
    'Timestamp': 'timestamp',
    'Which course are you currently enrolled in?': 'course_enrolled',
    'What degree program are you currently enrolled in?': 'degree_program',
    'On a scale of 1 to 5, with 1 being not at all interested and 5 being extremely interested, how interested are you taking more courses in Computer Science, Information Technology or Game Development?': 'computing_interest_score',
    'Gender': 'gender',
    'Race/ethnicity': 'race_ethnicity',
    'Age': 'age_group'
}


df = df.rename(columns=rename_dict)
df[['timestamp', 'course_enrolled', 'degree_program', 'computing_interest_score', 'gender', 'race_ethnicity', 'age_group']].head()
```

### 4.2 Cleaning Redundant Values in Degree Programs
The raw `degree_program` column has 23 unique responses, including redundant entries (e.g., `'Nursing'`, `'child psychology'`, `'Biology - Environmental Track'`). We consolidate these into standard broader academic divisions:
* **Health Sciences:** Radiography, Nursing, etc.
* **Social Sciences:** Psychology, Child and Family Studies, Criminal Justice
* **Arts & Communication:** Broadcasting, Journalism, Musical Theatre, Communication, Music Technology
* **STEM & Engineering:** Biology, Exercise Science, Engineering Science
* **Liberal Arts & General Education:** Liberal Arts, Non-degree students

```python
print("--- Before Cleaning ---")
print(df['degree_program'].value_counts())

def categorize_degree(degree):
    if pd.isna(degree):
        return 'Other'
    deg = str(degree).strip().lower()
    if any(k in deg for k in ['radiography', 'nursing']):
        return 'Health Sciences'
    elif any(k in deg for k in ['child', 'psychology', 'criminal justice']):
        return 'Social Sciences'
    elif any(k in deg for k in ['broadcasting', 'journalism', 'theatre', 'music', 'communication']):
        return 'Arts & Communication'
    elif any(k in deg for k in ['biology', 'exercise']):
        return 'STEM'
    elif 'engineering' in deg:
        return 'Engineering'
    elif 'liberal arts' in deg:
        return 'Liberal Arts'
    elif 'culinary' in deg or 'fire' in deg or 'history' in deg or 'analyst' in deg:
        return 'Other Applied Sciences/Humanities'
    else:
        return 'Not in Degree Program'

df['degree_category'] = df['degree_program'].apply(categorize_degree)

print("\n--- After Cleaning (Categorized) ---")
df['degree_category'].value_counts()
```

### 4.3 Standardizing Demographic Categories
Standardize multi-selection values in `race_ethnicity` to consolidate multi-racial combinations into a unified `'Multi-Racial'` category for clearer visualizations and reporting.

```python
def categorize_race(race):
    if pd.isna(race):
        return 'Unknown'
    if ';' in str(race):
        return 'Multi-Racial'
    return str(race).strip()

df['race_ethnicity_clean'] = df['race_ethnicity'].apply(categorize_race)

# Display updated demographic counts
df['race_ethnicity_clean'].value_counts()
```
